```python
from pathlib import Path 
import pandas as pd

#find the path to the data file
Spotify = Path("/Users/helenebg/Library/CloudStorage/OneDrive-UiTOffice365/Forskerlinje/Hel-8048/Eksamen/Spotify-analysis/dataset.csv")
#Read spotify data
df = pd.read_csv(Spotify)

#Showing the first 5 rows of the dataframe
print(df.head())
```

```python
#Function to clean the Spotify dataset

def tweak_df(df):
    return (
        df
    
    #Select columns needed for analysis
    .loc[:, ["track_id",
             "track_name", 
             "artists",
             "popularity",
             "valence",
             "track_genre"
         ]]

    #remove rowes with missing values
    .dropna()

    #keep values within the expected range for valence and popularity
    .query("popularity >= 0 and popularity <= 100 and valence >= 0 and valence <= 1")

    #set appropriate data types
    .astype({
        "track_name": "string",
        "artists": "string",
        "track_genre": "category",
        })
    )

#Apply the function to clean the dataset
music = tweak_df(df)

#Show the first 5 rows of the cleaned dataset
print(music.head())
```

```python
#compare the shape of the original and cleaned datasets
print("Original dataset:", len(df), "rows")

print("Cleaned dataset:", len(music), "rows")

print("Rows removed:", len(df) - len(music))

# Check for duplicate songs

print("Duplicate rows:", music.duplicated().sum())

print("Duplicate track IDs:",
      music["track_id"].duplicated().sum())

```
